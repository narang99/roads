```python
# extract the taco dataset to a directory in unblurred, alongside clean

EXTRACT_DEST_DIR = Path("/Users/hariomnarang/Desktop/personal/roads/datasets/test-samples/neg-masking/V1/rocks/classification/blurred/clean/taco")
TACO_SRC_DIR = TACO_PATH

areas_bbox_images = []


def get_level(path):
    return int(Path(path).stem.split("_")[0])

images = list((TACO_SRC_DIR / "images").glob("*.jpeg"))
for image in tqdm(images):
    mask = TACO_SRC_DIR / "masks" / f"{image.stem}.png"
    bboxes = list(get_region_crops(DiskBooleanMask.load(mask)))

    for bb in bboxes:
        area = bb.h * bb.w
        areas_bbox_images.append((area, bb, image))
```

```python
# 2. Extract the first number to define bins
# We create a bin for every 100 units (0-99 = Bin 0, 100-199 = Bin 1)
MIN_CLUSTER_AREA = 49283
areas_bbox_images = [item for item in areas_bbox_images if item[0] < MIN_CLUSTER_AREA and item[0] > 40]
areas = [item[0] for item in areas_bbox_images]
areas = np.array(areas)
bins = areas // 10
```

```python
plt.hist([a for a in areas if a <= 1000])
```

```python
import pandas as pd
# 1. Convert your list of tuples to a DataFrame
# Assuming: my_tuples = [(val, meta), (val, meta), ...]
df = pd.DataFrame(areas_bbox_images, columns=['area', 'bbox', 'image'])

# 2. Define your bins (e.g., by 100s)
df['bin'] = df['area'] // 10

# 3. Sample PROPORTIONALLY (e.g., 10% of the total data)
# frac=0.1 means "take 10% of each group"
sampled_df = df.groupby('bin', group_keys=False).apply(lambda x: x.sample(frac=0.02))

# 4. Convert back to your original format
final_sample = list(sampled_df[['area', 'bbox', 'image']].itertuples(index=False, name=None))
```

```python
from mtrain.utils import mkdir
imgs, masks = [], []

# first save the non clustered ones, sampled
maskdir, imagedir = mkdir(EXTRACT_DEST_DIR / "masks"), mkdir(EXTRACT_DEST_DIR / "train")

for i, (a, bb, image_path) in enumerate(tqdm(final_sample)):
    image_path = Path(image_path)
    image = DiskImage.load(image_path)
    mask_path = TACO_PATH / "masks" / f"{image_path.stem}.png"
    mask = DiskBooleanMask.load(mask_path)

    tight_img, new_y1, new_x1 = padded_crop(image, bb, 224)
    tight_mask = bbox_only_mask(mask, bb, 224)
    
    fname = f"trash_{image_path.stem}_{i}"

    DiskImage.save(tight_img, imagedir / f"{fname}.jpg")
    DiskBooleanMask.save(tight_mask, maskdir / f"{fname}.png")
```

```python
from pathlib import Path
from mtrain.example_dir.iterdir import get_dirs

DELHI = Path("/Users/hariomnarang/Desktop/personal/roads/datasets/inference/delhi_sample_500_results_test")
# ! ls {list(get_dirs(DELHI))[0]}
```

```python
dirs = list(get_dirs(DELHI))
```

```python
from mtrain.example_dir.core import load_npz
from tqdm import tqdm

def _get_trash_mask(name_by_paths, name):
    paths = name_by_paths[name]
    trash =load_npz(paths["trash"])
    other = load_npz(paths["other"])

    return trash > other

finals = []
for d in tqdm(dirs):
    edir = ExampleDir(d)
    try:
        name_by_paths = edir.negmask_100x100_paths()
        se224 = _get_trash_mask(name_by_paths, "step_edge_224")
        se = _get_trash_mask(name_by_paths, "step_edge")
        unb = _get_trash_mask(name_by_paths, "unblurred")

        # final_mask = np.zeros(se224.shape)
        final_mask = (se224 & ~se) | (se224 & ~unb) | (se & ~se224) | (se & ~unb) | (unb & ~se) | (unb & ~se224)

        DiskBooleanMask.save(final_mask, d / "negmask-miss.png")
    except Exception as ex:
        print(f"ERROR d={d} cause={ex}")
    # finals.append((edir.load_and_resize_image(), final_mask))
```
